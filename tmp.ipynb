```python
# From quiz 2.3

def RMSE(y, y_hat):
    y = np.array(y)
    y_hat = np.array(y_hat)
    d = (y-y_hat)**2
    MSE = np.mean(d)
    return np.sqrt(MSE)

def theta_optim(X,y):
    X = np.array(X)
    y = np.array(y).T #input is as row vector
    return np.linalg.inv(X.T @ X) @ X.T @ y

def empirical_risk_MSE(X, y, theta):
    X = np.array(X)
    theta = np.array(theta).T #input is a row vector
    y_hat = X @ theta
    diff = (y - y_hat)**2
    return np.mean(diff) 
    
```

```python
y = [7,2,7]
y_hat = [7,0,4]
res = RMSE(y,y_hat)
print(res)
```
